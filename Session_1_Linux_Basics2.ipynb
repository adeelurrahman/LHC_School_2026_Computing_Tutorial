## 📍 Module 1: What is Linux & UNIX Architecture?

### 📖 Key Concepts
* **Operating System (OS):** System software that acts as a resource manager and interface between the user and hardware.
* **Linux:** An open-source, multi-tasking, multi-user OS created by **Linus Torvalds** (1991) combined with **GNU utilities** created by **Richard Stallman** ($\text{Linux} + \text{GNU} = \text{Free UNIX}$).
* **Core Constituents:**
  1. **Kernel:** The central core that manages resources (memory, CPU, disk storage).
  2. **Shell:** The command interpreter between the user and kernel (e.g., `bash`, `csh`, `zsh`).
  3. **File System:** Hierarchical directory tree where **everything is treated as a file**.

## 📍 Module 3: File Operations & Text Processing

Commands allow you to manipulate data and inspect logs.

* **Creating & Copying:** `mkdir`, `touch`, `cp`, `mv`, `rm`.
* **Viewing Content:** `cat`, `echo`, `head`, `tail`, `grep`.
* **Counting & Sorting:** `wc` (word/line count), `sort`.

```python
%%bash
# Cell 3: Create files and search for patterns using wildcards and grep
cd lhc_workspace

# Generate dummy physics log files
echo "Event 1: Muon pT = 45.2 GeV" > run_01.log
echo "Event 2: Electron pT = 12.8 GeV" >> run_01.log
echo "Event 3: Muon pT = 62.1 GeV" >> run_01.log

echo "--- Content of run_01.log ---"
cat run_01.log

echo -e "\n--- Searching for 'Muon' events with grep ---"
grep "Muon" run_01.log

echo -e "\n--- Line count using wc ---"
wc -l run_01.log
```

## 📍 Module 4: Linux File Permissions

Permissions are represented in a **10-bit format** from `ls -l`:
* **Bit 1:** File type (`-` for file, `d` for directory).
* **Bits 2–4:** Owner permissions.
* **Bits 5–7:** Group permissions.
* **Bits 8–10:** Others permissions.

| Permission | Representation | Octal Value |
| :--- | :--- | :--- |
| **Read** | `r` | **4** |
| **Write** | `w` | **2** |
| **Execute** | `x` | **1** |

> **Octal Example:** `rwxrw-r--` $= (4+2+1) \ (4+2+0) \ (4+0+0) = \mathbf{764}$.

```python
%%bash
# Cell 4: Changing file permissions using chmod
cd lhc_workspace
echo '#!/bin/bash' > test_script.sh
echo 'echo "Hello from Linux Shell Script!"' >> test_script.sh

# Make the script executable (755 = rwxr-xr-x)
chmod 755 test_script.sh
ls -l test_script.sh

# Run the script
./test_script.sh
```

## 📍 Module 5: Processes & Environment Variables

Everything in Linux is either a **file** or a **process**.
* **Process Management:** `ps` (report snapshot), `kill -9 <PID>` (terminate process).
* **Environment Variables:** Global variables controlling the shell behavior.
  * `$HOME`: Your home directory.
  * `$PATH`: Directory search path for executables.
  * `$LD_LIBRARY_PATH`: Search path for dynamic libraries (crucial for ROOT/CMSSW).

```python
%%bash
# Cell 5: Inspecting environment variables and active processes
echo "User Home Directory: $HOME"
echo "Current Shell: $SHELL"
echo "Active Terminal Type: $TERM"

echo -e "\n--- Process Snapshot (ps) ---"
ps -ef | head -n 5
```
