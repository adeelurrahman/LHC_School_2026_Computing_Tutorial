# Part 1: Core Syntax, Data Structures, and Control Flow

## 1. Basics & Formatting
Python relies on whitespace and indentation to define logical code blocks, rather than curly braces `{}` or keywords.

```python
# Demonstrating literal constants, string formatting, and logical lines
i = 5
print('Value is %d' % i)

# Explicit line joining using backslash (\)
s = 'This is a multi-line string. \
This continues the string.'
print(s)
```

## 2. Control Flow: `if`, `while`, and `for`

Control flow statements guide execution. Note that loops in Python can feature an optional `else` block executed when the loop completes normally.

```python
# If statement
number = 23
guess = 23

if guess == number:
    print('Congratulations, you guessed it.')
elif guess < number:
    print ('No, it is a little higher than that')
else:
    print ('No, it is a little lower than that')

# While loop with else clause
count = 0
while count < 3:
    print ('Count is:', count)
    count += 1
else:
    print ('The while loop is over.')

# For loop over a range
for i in range(1, 5):
    print (i)
else:
    print ('The for loop is over.')
```

## 3. Basic Data Structures: Lists, Tuples, Dictionaries, and Slices

```python
# Lists (mutable sequence)
shoplist = ['apple', 'mango', 'carrot', 'banana']
shoplist.append('rice')
shoplist.sort()
print ('Sorted shopping list:', shoplist)

# Tuples (immutable sequence)
zoo = ('python', 'elephant', 'penguin')
print ('Number of animals in the zoo:', len(zoo))

# Dictionaries (key-value mapping)
ab = {
    'Swaroop': 'swaroop@swaroopch.com',
    'Larry': 'larry@wall.org',
    'Matsumoto': 'matz@ruby-lang.org'
}
print ("Swaroop's address is %s" % ab['Swaroop'])

# Slicing operations on lists and tuples
print ('Item 1 to 3:', shoplist[1:3])
print ('Item start to end:', shoplist[:])
```

# Part 2: Functions, Module Structure, File I/O, and Data Manipulation

## 1. Advanced Functions and Docstrings
Functions accept default and keyword arguments, and support runtime documentation strings (`__doc__`).

```python
def printMax(x, y):
    '''Prints the maximum of two numbers.

    The two values must be integers.'''
    x = int(x)
    y = int(y)
    if x > y:
        print (x, 'is maximum')
    else:
        print (y, 'is maximum')

printMax(3, 5)
print (printMax.__doc__)

# Default arguments and keyword arguments
def say(message, times=1):
    print (message * times)

say('Hello ')
say('World ', times=3)

# Lambda expressions (anonymous functions)
double = lambda x: x * 2
print ('Double of 5 is:', double(5))
```

## 2. Dynamic List Manipulation & Stack/Queue Paradigms

```python
# Using a list as a LIFO Stack
stack = [3, 4, 5]
stack.append(6)
stack.append(7)
print ('Stack after pushes:', stack)
print ('Popped element:', stack.pop())
print ('Stack remaining:', stack)

# Deleting slice elements
a = [-1, 1, 66.25, 333, 333, 1234.5]
del a[2:4]
print ('List after del a[2:4]:', a)
```

## 3. File Input/Output & Object Serialization (`pickle`)
Files can be written to, read from, and Python objects can be serialized to disk using `cPickle` or `pickle`.

```python
import pickle

# Writing text to file
poem = '''\
Programming is fun
When the work is done
if you want to make your work also fun:
    use Python!
'''
with open('poem.txt', 'w') as f:
  f.write(poem)
f.close()

# Reading text from file
with open('poem.txt', 'r') as f:
  while True:
    line = f.readline()
    if len(line) == 0:
        break
    print (line,)
f.close()

# Object serialization via pickle
shoplistfile = 'shoplist.data'
shoplist = ['apple', 'mango', 'carrot']

with open(shoplistfile, 'wb') as f:
  pickle.dump(shoplist, f)
f.close()

with open(shoplistfile, 'rb') as f:
  storedlist = pickle.load(f)
print ('Deserialized list:', storedlist)
f.close()
```

# Part 3: Object-Oriented Programming, Advanced Features, Exceptions, and GUIs

## 1. Object-Oriented Programming (OOP) & Inheritance
Classes combine data and behavior. The `self` parameter explicitly references the running instance.

```python
class Person:
    '''Represents a person.'''
    population = 0  # Class variable

    def __init__(self, name):
        self.name = name  # Instance variable
        Person.population += 1

    def sayHi(self):
        print ('Hi, my name is %s.' % self.name)

p1 = Person('Alice')
p1.sayHi()
print ('Total population:', Person.population)

# Inheritance
class SchoolMember:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Teacher(SchoolMember):
    def __init__(self, name, age, salary):
        SchoolMember.__init__(self, name, age)
        self.salary = salary

    def tell(self):
        print ('Name: "%s" Age: "%s" Salary: "%s"' % (self.name, self.age, self.salary))

t = Teacher('Dr. Smith', 40, 30000)
t.tell()
```

## 2. Iterators and Generators
Iterators implement `__iter__()` and `next()`. Generators streamline iterator creation using the `yield` statement.

```python
# Custom Iterator Class
class Reverse:
    "Iterator for looping over a sequence backwards"
    def __init__(self, data):
        self.data = data
        self.index = len(data)

    def __iter__(self):
        return self

    def __next__(self):
        if self.index == 0:
            raise StopIteration
        self.index -= 1
        return self.data[self.index]

rev = Reverse('spam')
for char in rev:
    print (char,)
print ('')

# Generator Function
def reverse_gen(data):
    for index in range(len(data)-1, -1, -1):
        yield data[index]

for char in reverse_gen('golf'):
    print (char,)
print ('')
```

## 3. Exception Handling
Exceptions gracefully handle errors using `try...except...else...finally` blocks.

```python
class ShortInputException(Exception):
    '''A user-defined exception class.'''
    def __init__(self, length, atleast):
        Exception.__init__(self)
        self.length = length
        self.atleast = atleast

try:
    text = 'hi'  # Length is 2
    if len(text) < 3:
        raise ShortInputException(len(text), 3)
except ShortInputException as ex:
    print ('ShortInputException: The input was length %d, expected at least %d' % (ex.length, ex.atleast))
else:
    print ('No exception was raised.')
finally:
    print ('Cleanup actions executed in finally block.')
```
