## 1. Introduction to C++

### What is a Computer Program?
A set of instructions written to perform a specific task on a system:
- Reading files from media
- Carrying out number-crunching scientific jobs
- Managing network chat sessions

A program serves as an interface between its user and the underlying hardware.

### What is C++?
C++ is a **statically typed**, **compiled**, **general-purpose**, **case-sensitive**, and **free-form** programming language designed as an enhancement to the C language.

It supports:
- Systems programming & direct hardware mapping
- Object-Oriented Programming (classes & inheritance)
- Programming-in-the-large (namespaces, exceptions)
- Generic programming (templates)
- High reusability via standard libraries

## 2. Life-Cycle of a C/C++ Program

1. **Source Code (`.cpp`, `.cc`):** Raw code written by the developer.
2. **Preprocessor:** Inlines `#include` directives, expands macros, and resolves preprocessor code.
3. **Compiler:** Translates source code into machine code and object files (`.o`, `.obj`).
4. **Linker:** Combines object files with external dynamic/static libraries (e.g., `libc.so`) to generate an **Executable**.

### Control Flow & Decision Making

````python
#include <iostream>
using namespace std;

int main() {
    int a = 20;
    
    // If-Else Branching
    if (a < 20) {
        cout << "a is less than 20" << endl;
    } else if (a > 20) {
        cout << "a is greater than 20" << endl;
    } else {
        cout << "a is 20" << endl;
    }

    // Switch Statement
    switch (a) {
        case 10:
            cout << "a is 10" << endl;
            break;
        case 20:
            cout << "Switch matched: a is 20!" << endl;
            break;
        default:
            cout << "Default case triggered" << endl;
    }
    return 0;
}
````

### Functions & Parameter Passing Mechanisms

C++ supports three distinct parameter passing mechanisms:
1. **Pass by Value:** Makes a copy. Suitable for small primitives where original data shouldn't be altered.
2. **Pass by Reference:** Shares variable reference directly. Suitable when modifications need to persist outside the function scope.
3. **Pass by Constant Reference (`const T&`):** Hybrid method. Efficient for large structures/objects without incurring copy overhead or allowing modification.

````python
#include <iostream>
using namespace std;

void passByValue(int i) {
    i = i * i;
}

void passByReference(int &i) {
    i = i * i;
}

int passByConstReference(const int &i) {
    return i * i;
}

int main() {
    int x = 5;
    passByValue(x);
    cout << "After Pass by Value: " << x << endl; // Remains 5

    passByReference(x);
    cout << "After Pass by Reference: " << x << endl; // Modifies to 25

    cout << "Const Reference Result: " << passByConstReference(x) << endl;
    return 0;
}
````

### Arrays & Pointers

````python
#include <iostream>
using namespace std;

int main() {
    // Array Initialization
    int foo[] = {20, 40, 60, 80, 100};
    int sum = 0;
    for (int n = 0; n < 5; ++n) {
        sum += foo[n];
    }
    cout << "Array Sum: " << sum << endl;

    // Pointer Basics
    int var1 = 11;
    int var2 = 22;
    int* ptr = nullptr;

    ptr = &var1; // Pointing to var1
    cout << "Address: " << ptr << " | Dereferenced Value: " << *ptr << endl;

    ptr = &var2; // Pointing to var2
    cout << "Address: " << ptr << " | Dereferenced Value: " << *ptr << endl;

    return 0;
}
````

## 4. Object-Oriented Programming (OOP) in C++

OOP in C++ rests on three foundational pillars:
1. **Encapsulation:** Binding data and functions into classes (`public`, `private`, `protected`).
2. **Inheritance:** Parent-child hierarchical relationships for code reuse.
3. **Polymorphism:** Supporting uniform interfaces across varying types (`virtual` functions, overloading).

````python
#include <iostream>
using namespace std;

// Base Class
class Shape {
protected:
    int xCo, yCo;
public:
    Shape(int x = 0, int y = 0) : xCo(x), yCo(y) {}
    virtual void whoami() {
        cout << "I am a generic Shape." << endl;
    }
    virtual ~Shape() {} // Virtual destructor for dynamic memory safety
};

// Derived Class: Circle
class Circle : public Shape {
private:
    float radius;
public:
    Circle(int x, int y, float rad) : Shape(x, y), radius(rad) {}
    void whoami() override {
        cout << "I am a Circle with radius: " << radius << endl;
    }
};

// Derived Class: Rectangle
class Rectangle : public Shape {
private:
    float length, width;
public:
    Rectangle(int x, int y, float l, float w) : Shape(x, y), length(l), width(w) {}
    void whoami() override {
        cout << "I am a Rectangle with dimensions: " << length << "x" << width << endl;
    }
};

int main() {
    Shape* pShape[2];
    pShape[0] = new Circle(10, 10, 5.5);
    pShape[1] = new Rectangle(5, 5, 20.0, 10.0);

    // Polymorphic invocation
    for (int i = 0; i < 2; ++i) {
        pShape[i]->whoami();
    }

    // Cleanup
    for (int i = 0; i < 2; ++i) {
        delete pShape[i];
    }
    return 0;
}
````

## 5. Generics & Standard Template Library (STL)

The STL provides standard components including:
- **Containers:** `vector`, `list`, `map`, `deque`, `set`
- **Iterators:** Generalized pointers used to traverse containers
- **Algorithms:** Common routines (`sort`, `find`, `copy`, `transform`)

````python
#include <iostream>
#include <list>
using namespace std;

int main() {
    list<char> lst;

    // Populate list container
    for (char ch = 'A'; ch <= 'Z'; ++ch) {
        lst.push_back(ch);
    }

    // Iterate and display elements using constant iterator
    list<char>::const_iterator pos;
    cout << "STL List contents: ";
    for (pos = lst.begin(); pos != lst.end(); ++pos) {
        cout << *pos << ' ';
    }
    cout << endl;
    return 0;
}
````
